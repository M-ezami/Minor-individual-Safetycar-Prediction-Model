```python
import pandas as pd

df = pd.read_csv("race_state_with_predictions.csv")
print(df.head())        # show first few rows
print(df.shape)         # show number of rows and columns
print(df.info())        # show column types and non‑null counts
```

```python
# Set probability threshold for classifying a lap as predicted incident
threshold = 0.2
# Create binary prediction column
df['Predicted'] = (df['IncidentProb'] > threshold).astype(int)

# False positives = predicted=1 but actual=0
false_alarms = df[(df['Predicted'] == 1) & (df['Target'] == 0)]

# Summary
print("Total rows:", len(df))
print("Total predicted incidents:", df['Predicted'].sum())
print("Total actual incidents:", df['Target'].sum())
print("Number of false alarms:", len(false_alarms))
print("Percentage of false alarms:", len(false_alarms)/len(df) * 100, "%")
```
